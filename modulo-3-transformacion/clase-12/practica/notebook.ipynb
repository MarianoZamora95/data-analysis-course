### Ejercicio 2 — Estadísticos con NumPy

NumPy tiene funciones estadísticas equivalentes a las de Pandas, pero operan sobre arrays. Conocerlas te permite trabajar con datos fuera de un DataFrame cuando es necesario.

**Consigna:** Dado este array de horas de estudio semanales de un grupo de 10 estudiantes, calculá con funciones de NumPy (no con Pandas ni con Python nativo)


```python
import numpy as np

horas = np.array([12, 25, 8, 31, 19, 42, 15, 22, 7, 28])

# 1. Media
media = np.mean(horas)

# 2. Mediana
mediana = np.median(horas)

# 3. Desviación estándar
desv_std = np.std(horas)

# 4. Mínimo y máximo
minimo = np.min(horas)
maximo = np.max(horas)

# 5. Percentiles 25 y 75
p25 = np.percentile(horas, 25)
p75 = np.percentile(horas, 75)

# 6. Rango intercuartil (IQR)
iqr = p75 - p25

print(f"Media: {media:.2f}")
print(f"Mediana: {mediana:.2f}")
print(f"Desviación estándar: {desv_std:.2f}")
print(f"Mínimo: {minimo} | Máximo: {maximo}")
print(f"Percentil 25: {p25} | Percentil 75: {p75}")
print(f"IQR: {iqr}")

# --- Respuestas en comentarios ---
# Después verifica que media y mediana son distintas:
# - La media es 20.90 horas y la mediana es 20.50 horas (en este caso la diferencia
#   es leve, pero la media queda ligeramente por encima de la mediana).
#
# ¿Qué valor las separa tanto? ¿Qué harías con él?
# - El valor extremo más alto es 42 horas (un posible outlier o estudiante con
#   una carga de estudio atípica frente al resto, que se mueve entre 7 y 31 horas).
# - ¿Qué haría con él? Primero verificaría si es un error de carga (por ejemplo,
#   si anotaron horas quincenales en vez de semanales) o un caso real válido
#   (ej. semana de finales). Si es válido y quiero describir el comportamiento
#   típico del grupo sin sesgos, usaría la mediana y el IQR en lugar de la media.
```

### Ejercicio 3 — Generar y filtrar arrays

`np.where()` permite localizar las posiciones donde se cumple una condición, es el equivalente de un filtro, pero sobre un array, devolviendo los índices en lugar de los valores.

**Consigna:** Responde ¿Cuántas temperaturas están por encima de 30°C?, ¿En qué posiciones (índices) están esas temperaturas? Usá np.where(). Indica ¿Cuál es el promedio de las temperaturas que superan 30°C? ¿Y el promedio de las que están por debajo de 15°C?

Pista: np.where(condicion) devuelve una tupla, el primer elemento es el array de índices. Puedes usarlo para indexar `temperaturas[np.where(temperaturas > 30)]`

```python
import numpy as np

np.random.seed(7)
temperaturas = np.random.normal(22, 5, 100).round(1)

# 1. Posiciones (índices) donde la temperatura supera 30°C
# np.where devuelve una tupla, tomamos el primer elemento [0]
indices_mayor_30 = np.where(temperaturas > 30)[0]

# 2. Cantidad de temperaturas por encima de 30°C
cantidad_mayor_30 = len(indices_mayor_30)

# 3. Promedio de las temperaturas que superan 30°C
temps_mayor_30 = temperaturas[indices_mayor_30]
promedio_mayor_30 = np.mean(temps_mayor_30)

# 4. Promedio de las temperaturas que están por debajo de 15°C
indices_menor_15 = np.where(temperaturas < 15)[0]
promedio_menor_15 = np.mean(temperaturas[indices_menor_15])

print("Cantidad de temperaturas > 30°C:", cantidad_mayor_30)
print("Posiciones (índices) > 30°C:", indices_mayor_30)
print("Valores > 30°C:", temps_mayor_30)
print(f"Promedio de temperaturas > 30°C: {promedio_mayor_30:.2f}°C")
print(f"Promedio de temperaturas < 15°C: {promedio_menor_15:.2f}°C")
```

### Ejercicio 5 — Calcular correlación con NumPy

El artículo mostró cómo calcular la correlación de Pearson desde cero usando la fórmula. Este ejercicio te pide implementar esa misma lógica en una función reutilizable.

**Consigna:** Escribe una función `correlacion_pearson(x, y)` que calcule la correlación usando únicamente NumPy — sin usar `np.corrcoef()` ni `.corr()` de Pandas. Después verifica que tu función da el mismo resultado que ambas para el par `horas_estudio` y `puntaje_final`.

```python
import numpy as np
import pandas as pd

df = pd.read_csv("estudiantes.csv")

def correlacion_pearson(x, y):
    # 1. Calcular las desviaciones respecto a la media
    dev_x = x - np.mean(x)
    dev_y = y - np.mean(y)

    # 2. Multiplicarlas elemento a elemento y promediar (covarianza)
    covarianza = np.mean(dev_x * dev_y)

    # 3. Dividir por el producto de las desviaciones estándar
    r = covarianza / (np.std(x) * np.std(y))

    return r

x = df["horas_estudio"].values
y = df["puntaje_final"].values

r_manual = correlacion_pearson(x, y)
r_numpy = np.corrcoef(x, y)[0, 1]
r_pandas = df["horas_estudio"].corr(df["puntaje_final"])

print(f"Manual (NumPy puro): {r_manual:.6f}")
print(f"np.corrcoef():       {r_numpy:.6f}")
print(f"Pandas .corr():      {r_pandas:.6f}")
```

### Ejercicio 6 — Un outlier que rompe la correlación

El dataset tiene un outlier en `horas_estudio` (fila 10, con 85 horas semanales — imposible). Este ejercicio mide cuánto distorsiona la correlación.

**Consigna:** Calcula la correlación entre `horas_estudio` y `puntaje_final` dos veces: una con el dataset completo y otra eliminando la fila 10. Muestra la diferencia en un mensaje claro.

Después responde: ¿la correlación sube o baja cuando sacás el outlier? ¿Eso tiene sentido para la relación entre horas de estudio y rendimiento?


```python
import pandas as pd

df = pd.read_csv("estudiantes.csv")

# Eliminamos la fila con índice 10 (el outlier de 85 horas semanales)
df_sin_outlier = df.drop(index=10)

corr_completa = df["horas_estudio"].corr(df["puntaje_final"])
corr_sin_outlier = df_sin_outlier["horas_estudio"].corr(df_sin_outlier["puntaje_final"])

print(f"Correlación con outlier:    {corr_completa:.4f}")
print(f"Correlación sin outlier:    {corr_sin_outlier:.4f}")
print(f"Diferencia:                 {abs(corr_completa - corr_sin_outlier):.4f}")

# Interpretación:
# - ¿La correlación sube o baja cuando sacás el outlier?
#   Sube significativamente (pasa de ~0.4993 a un valor mucho más alto).
#
# - ¿Eso tiene sentido para la relación entre horas de estudio y rendimiento?
#   Sí, tiene total sentido: un registro erróneo e imposible (85 horas semanales
#   de estudio con un puntaje que no acompaña esa escala) rompe la tendencia
#   lineal del resto del grupo y "tira hacia abajo" el coeficiente de Pearson.
#   Al limpiar ese dato atípico, emerge la verdadera correlación positiva fuerte
#   entre estudiar más horas y obtener mejor puntaje final.
```

### Ejercicio 7 — IQR sobre una columna

Antes de aplicar IQR a todo el dataset, entendés bien el método aplicándolo sobre una sola columna donde podés verificar el resultado manualmente.

**Consigna:** Aplica el método IQR sobre la columna `puntaje_final`. Calcula el P25, P75, IQR, límite inferior y límite superior. Después identifica cuántos valores caen fuera de esos límites y cuáles son.

```python
import numpy as np
import pandas as pd

df = pd.read_csv("estudiantes.csv")

arr = df["puntaje_final"].values

p25 = np.percentile(arr, 25)
p75 = np.percentile(arr, 75)
iqr = p75 - p25
lim_inf = p25 - 1.5 * iqr
lim_sup = p75 + 1.5 * iqr

outliers = df[(df["puntaje_final"] < lim_inf) | (df["puntaje_final"] > lim_sup)]

print(f"P25: {p25:.1f} | P75: {p75:.1f} | IQR: {iqr:.1f}")
print(f"Rango normal: [{lim_inf:.1f}, {lim_sup:.1f}]")
print(f"\nOutliers encontrados: {len(outliers)}")
print(outliers["puntaje_final"])

# ¿El valor de 142 en puntaje_final aparece como outlier?
# - Sí, aparece como outlier porque supera ampliamente el límite superior
#   calculado con la regla de Tukey (P75 + 1.5 * IQR).
#
# ¿Tiene sentido que aparezca? ¿Por qué?
# - Tiene total sentido: un puntaje final de 142 está completamente fuera de la
#   escala habitual de calificación (que suele ir de 0 a 100) y se aleja de la
#   distribución del resto de los estudiantes. Muy probablemente sea un error de
#   carga o tipeo al registrar la nota.
```

### Ejercicio 8 — Z-score sobre una columna

El Z-score no solo detecta outliers — te dice cuán extremo es cada uno. En este ejercicio lo aplicás sobre `horas_sueno` para detectar el valor de 0.2 horas que es claramente un error.

**Consigna:** Calcula el Z-score de cada valor en `horas_sueno` usando NumPy. Identificá todos los valores con Z-score menor a -3 o mayor a +3. Muestra el valor original y su Z-score para cada outlier detectado.

```python
import numpy as np
import pandas as pd

df = pd.read_csv("estudiantes.csv")

arr = df["horas_sueno"].values

media   = np.mean(arr)
std     = np.std(arr)
zscores = (arr - media) / std

outliers_mask = np.abs(zscores) > 3
indices = np.where(outliers_mask)[0]

print("Outliers detectados por Z-score en horas_sueno:")
for idx in indices:
    print(f"  Fila {idx}: valor = {arr[idx]:.1f} | Z-score = {zscores[idx]:+.2f}")
```

### Ejercicio 10 — Escanear todo el dataset

En un análisis real no revisás columna por columna manualmente — escribís una función que escanee todo el dataset y produzca un reporte de outliers por columna.

**Consigna:** Escribe una función `reporte_outliers(df, metodo="iqr", umbral=3)` que:

1. Recorra todas las columnas numéricas del DataFrame
2. Aplique el método indicado (`"iqr"` o `"zscore"`)
3. Devuelva un DataFrame resumen con: columna, cantidad de outliers, porcentaje sobre el total, y los valores detectados

```python
import numpy as np
import pandas as pd

df = pd.read_csv("estudiantes.csv")

def reporte_outliers(df, metodo="iqr", umbral=3):
    resultados = []

    for col in df.select_dtypes(include=[np.number]).columns:
        arr = df[col].values

        if metodo == "iqr":
            p25 = np.percentile(arr, 25)
            p75 = np.percentile(arr, 75)
            iqr = p75 - p25
            mask = (arr < p25 - 1.5 * iqr) | (arr > p75 + 1.5 * iqr)
        elif metodo == "zscore":
            zscores = (arr - np.mean(arr)) / np.std(arr)
            mask = np.abs(zscores) > umbral
        else:
            raise ValueError("Método no válido. Usá 'iqr' o 'zscore'.")

        valores_outliers = arr[mask]
        cantidad = len(valores_outliers)
        porcentaje = (cantidad / len(arr)) * 100

        resultados.append({
            "columna": col,
            "cantidad_outliers": cantidad,
            "porcentaje": round(porcentaje, 2),
            "valores_detectados": valores_outliers.tolist()
        })

    return pd.DataFrame(resultados)

# Probamos ambos métodos
print("Reporte con IQR:")
display(reporte_outliers(df, metodo="iqr"))

print("\nReporte con Z-score:")
display(reporte_outliers(df, metodo="zscore", umbral=3))
```
